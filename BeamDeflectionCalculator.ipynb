```python
import tkinter as tk
from tkinter import ttk
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
```

#Creat GUI Window

```python
window_beam = tk.Tk()
```

```python
window_beam.title("Beam Deflection Calculator")
```

```python
pad_x = 10
pad_y = 10
Length = ttk.Label(window_beam, text="Length of the beam (L_m):")
Length.grid(row=0, column=0, padx=pad_x, pady=pad_y)
```

```python
L_entry = ttk.Entry(window_beam)
L_entry.grid(row=0, column=1, padx=0, pady=0)
```

```python
Load = ttk.Label(window_beam, text="Load (P_kN):")
Load.grid(row=1, column=0, padx=pad_x, pady=pad_y)
Load_entry = ttk.Entry(window_beam)
Load_entry.grid(row=1, column=1, padx=pad_x, pady=pad_y)
```
